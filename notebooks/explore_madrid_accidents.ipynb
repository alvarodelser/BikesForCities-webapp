# Exploración de Datos de Accidentes en Madrid

Este notebook analiza los valores únicos que pueden tomar los campos de `tipo_accidente`, `lesividad` (severidad) y `tipo_vehiculo` en los CSVs de origen de Madrid Open Data.

```python
import pandas as pd

# Cargar el archivo descargado
df_all = pd.read_csv('../data/madrid_accidents_2024.csv', sep=';', encoding='utf-8-sig', dtype=str)
df_all.columns = [c.lower().strip().replace(" ", "_") for c in df_all.columns]
print(f"Cargados {len(df_all)} registros.")
```

```python
# Función para mostrar valores únicos y sus frecuencias
def mostrar_valores_unicos(df, col):
    if col not in df.columns:
        print(f"❌ No se encontró la columna '{col}'")
        return
        
    print(f"📊 Valores para la columna '{col}':\n")
    # Agrupar, contar y ordenar
    conteo = df[col].fillna("NULO/VACÍO").value_counts()
    
    # Mostrar como tabla con porcentajes
    df_stats = pd.DataFrame({
        'Conteo': conteo,
        'Porcentaje (%)': (conteo / len(df) * 100).round(2)
    })
    
    display(df_stats)
```

## 1. Tipo de Accidente (`tipo_accidente`)

```python
mostrar_valores_unicos(df_all, 'tipo_accidente')
```

## 2. Severidad / Lesividad (`lesividad`)

```python
mostrar_valores_unicos(df_all, 'lesividad')
```

## 3. Tipo de Vehículo (`tipo_vehiculo`)

```python
mostrar_valores_unicos(df_all, 'tipo_vehiculo')
```

## 4. Tipos de Vehículo que contienen la palabra 'bici'

```python
df_bicis = df_all[df_all['tipo_vehiculo'].fillna('').str.lower().str.contains('bici')]
print(f"🚴 Registros implicando bicicletas: {len(df_bicis)}")
mostrar_valores_unicos(df_bicis, 'tipo_vehiculo')
```
